```python
import numpy as np
from matplotlib import pyplot  as plt
from matplotlib import pylab,rc
from matplotlib.ticker import AutoMinorLocator, MultipleLocator, FormatStrFormatter
import numpy as np
from numpy.random import normal
import sys
from ase.io import read
from ase.units import kB
import glob, os
import re
from ase.io import read
%matplotlib inline

from grsq.debye import Debye
```

# AB cis - AB trans

```python
path = './'
f_off = path + 'abt.xyz' 
f_on = path + 'abc.xyz' 
s_off = read(f_off)
s_on = read(f_on)
q = np.arange(0,10,0.01)
deb = Debye(qvec=q)
deb_ab_trans = deb.debye(s_off)
deb_ab_cis = deb.debye(s_on)
deb_diff_ab_cis_trans = deb_ab_cis - deb_ab_trans

fout = 'ds_ab.xyz'

data = np.column_stack((q, deb_diff_ab_cis_trans))
np.savetxt(fout, data)
```

```python
fontsize=20

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(9,6), sharey=True)

ax.plot(q, deb_diff_ab_cis_trans,linestyle = "-", linewidth = 3, marker = 'None', alpha = 1.0,label = 'AB')

ax.set_xlim([0.5,4])
# ax.set_ylim([-140,180])

ax.tick_params(direction='in', length=6, width=2)

plt.xticks(fontsize=fontsize)
plt.yticks(fontsize=fontsize)

ytext = r'$\Delta S$' 
ax.set_ylabel(ytext, fontweight='normal',fontsize = fontsize,labelpad = 15)
xtext = r'$q$ (1/Å)'
ax.set_xlabel(xtext, labelpad = 5, fontweight = 'normal',fontsize = fontsize)

ax.legend(loc='best', fontsize = 16)

fig.tight_layout()
plt.show()
```

# AB with different substituents

```python
path = './'
q = np.arange(0,10,0.01)
deb = Debye(qvec=q)
abt = read(path + 'abt.xyz')
abc = read(path + 'abc.xyz')
abts = read(path + 'abts.xyz')

abnt = read(path + 'abnt.xyz')
abnc = read(path + 'abnc.xyz')
abnts = read(path + 'abnts.xyz')

d2t = read(path + 'd2t.xyz')
d2c = read(path + 'd2c.xyz')
d2ts = read(path + 'd2ts.xyz')

deb_abt = deb.debye(abt) 
deb_abc = deb.debye(abc)
deb_abts = deb.debye(abts)

deb_abnt = deb.debye(abnt) 
deb_abnc = deb.debye(abnc)
deb_abnts = deb.debye(abnts)

deb_d2t = deb.debye(d2t) 
deb_d2c = deb.debye(d2c)
deb_d2ts = deb.debye(d2ts)
```

```python

deb_diff_abc_abt = deb_abc - deb_abt
deb_diff_abts_abt = deb_abts - deb_abt
deb_diff_abts_abc = deb_abts - deb_abc


deb_diff_abnc_abnt = deb_abnc - deb_abnt
deb_diff_abnts_abnt = deb_abnts - deb_abnt
deb_diff_abnts_abnc = deb_abnts - deb_abnc


deb_diff_d2c_d2t = deb_d2c - deb_d2t
deb_diff_d2ts_d2t = deb_d2ts - deb_d2t
deb_diff_d2ts_d2c = deb_d2ts - deb_d2c

# fout = 'ds_ab.xyz'

# data = np.column_stack((q, deb_diff_ab_ts_cis))
# np.savetxt(fout, data)
```

```python
fontsize=20

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(9,6), sharey=True)

ax.plot(q, deb_diff_abc_abt,linestyle = "-", linewidth = 3, marker = 'None', alpha = 1.0,label = 'AB')
ax.plot(q, deb_diff_abnc_abnt,linestyle = "-", linewidth = 3, marker = 'None', alpha = 1.0,label = 'AB-Br')
ax.plot(q, deb_diff_d2c_d2t,linestyle = "-", linewidth = 3, marker = 'None', alpha = 1.0,label = 'Dye 2')

ax.set_xlim([0.5,4])
ax.set_ylim([-1000,1000])

ax.tick_params(direction='in', length=6, width=2)

plt.xticks(fontsize=fontsize)
plt.yticks(fontsize=fontsize)

ytext = r'$\Delta S$' 
ax.set_ylabel(ytext, fontweight='normal',fontsize = fontsize,labelpad = 15)
xtext = r'$q$ (1/Å)'
ax.set_xlabel(xtext, labelpad = 5, fontweight = 'normal',fontsize = fontsize)

ax.legend(loc='best', fontsize = 16)

fig.tight_layout()
plt.show()
```
